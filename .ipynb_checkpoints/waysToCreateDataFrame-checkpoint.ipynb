```python
import pandas as pd

weather_data = (
    ["Monday", "Tuesday", "Wednesday"],   # days
    [30, 32, 28],                        # temperature
    [70, 65, 80]                         # humidity
)

df = pd.DataFrame(weather_data).T
df.columns = ["day", "temperature", "humidity"]

print(df)
```

```python
weather_data = [
    {"day": "Monday", "temperature": 30, "humidity": 70},
    {"day": "Tuesday", "temperature": 32, "humidity": 65}
]

df = pd.DataFrame(weather_data)
df
```
